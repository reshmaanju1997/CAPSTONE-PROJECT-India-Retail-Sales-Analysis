Handling Missing values

*Missing values were identified using df.isnull().sum() and handled appropriately based on the nature and importance of each column. Suitable techniques such as filling missing values with meaningful categories or statistical values were applied to improve data quality and ensure reliable analysis*

City , Order Priority , Sales Rep ID , Gift Wrap and warehouse column were handled with **'UNKNOWN'**
*   **City**  ----- The actual city was unavailable, so "Unknown" was used to preserve the record without assuming a location.
*   **Order Priority**  ----- The actual priority was not available, so "Unknown" avoids incorrectly classifying an order as Critical, High, Medium, or Low.
*   **Sales Rep ID**  ----- The responsible sales representative was unavailable, so "Unknown" preserves the transaction without assigning it to an incorrect salesperson.
*  **Gift wrap**  ----- The responsible sales representative was unavailable, so "Unknown" preserves the transaction without assigning it to an incorrect salesperson.
*   **Warehouse**  ----- The warehouse information was unavailable, so "Unknown" prevents assigning the order to an incorrect warehouse.


Missing values in the City, Order Priority, Sales Rep ID, Gift Wrap, and Warehouse Code columns were replaced with **"Unknown"**. These columns contain categorical or identifier information, and the actual missing values could not be reliably determined from the available data. Therefore **"Unknown"** was used to preserve the records, avoid introducing incorrect assumptions, and maintain the integrity of the dataset for further analysis.


Customer Feedback score  ----- the Customer Feedback Score contains rating values such as 1, 2, 3, 4, and 5. For missing values, using the median.

Why Median?
Missing values were imputed using the **median** of the available feedback scores. The median was selected because customer ratings are bounded on a 1–5 scale and the median provides a robust measure of the central tendency without being strongly influenced by extreme ratings.

**Mean** was not preferred because it can produce decimal values and can be affected by the distribution of ratings

**Mode** could over-represent the most frequently occurring rating.

Convert Data Types

*Data type correction is the process of converting columns from inappropriate or inconsistent data types into suitable data types based on the nature of the data. This step is important because incorrect data types can cause errors during data analysis, calculations, filtering, grouping, and visualization.*

```python
#order date and ship date
#change datatype

df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    dayfirst=True,
    errors='coerce')
df['Ship Date'] = pd.to_datetime(
    df['Ship Date'],
    dayfirst=True,
    errors='coerce')
```

```python
#verify

df[['Order Date', 'Ship Date']].dtypes
```

```python
#list price

#Remove Commas
df['List Price'] = df['List Price'].astype(str).str.replace(',', '', regex=False)

#Convert to numeric
df['List Price'] = pd.to_numeric(df['List Price'], errors='coerce')

#Rounded the decimal value
df['List Price'] = df['List Price'].round(0)

#Check datatypes
df['List Price'] = df['List Price'].astype('Int64')

#Check first 13 rows
print(df['List Price'].head(13))
```

```python
#Sales

# Remove commas
df['Sales'] = df['Sales'].astype(str).str.replace(',', '', regex=False)

# Convert to numeric
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')

# Round to 2 decimal places
df['Sales'] = df['Sales'].round(2)

# Check datatype
print(df['Sales'].dtype)

#check first 13 rows
print(df['Sales'].head(13))
```

# Feature Engineering

*Feature engineering is the process of creating new meaningful features (columns) from existing data to make the dataset more useful for analysis and to uncover additional patterns and insights.*

Derived 6 Columns

| Derived Feature          | Explanation                                                                                               | Why it was created                                                                             |
| ------------------------ | --------------------------------------------------------------------------------------------------------- | ---------------------------------------------------------------------------------------------- |
| **Shipping Days**        | Calculates the number of days between the Order Date and Ship Date.                                       | To measure shipping duration and understand delivery efficiency.                               |
| **Order Year**           | Extracts the year from the Order Date.                                                                    | To analyze yearly sales trends and compare business performance across years.                  |
| **Order Month**          | Extracts the month from the Order Date.                                                                   | To identify monthly sales patterns, seasonal trends, and high/low sales periods.               |
| **Sales per Quantity**   | Calculates sales amount per unit by dividing Sales by Quantity.                                           | To understand the average sales value generated per product unit.                              |
| **Shipping Performance** | Categorizes orders based on the number of Shipping Days, such as Fast, Standard, or Slow.                 | To evaluate shipping efficiency and identify orders that may require attention.                |
| **Returned Flag**        | Converts the Returned information into a simplified flag indicating whether an order was returned or not. | To make return analysis easier and identify return patterns and their relationship with sales. |


1.Shipping Days

```python
#Number of days between order date and ship date

df['Shipping Days'] = (
    df['Ship Date'] - df['Order Date']).dt.days.astype('Int64')
```

2.Order Year

```python
#Extract year from the order date

df['Order Year'] = df['Order Date'].dt.year
```

3. Order Month

```python
#Extract Month From the order date

df['Order Month'] = df['Order Date'].dt.month
```

4. Sales per Quantity

```python
#Calculate Sales per unit

df['Sales per Quantity'] = (
    df['Sales'] / df['Quantity'])
```

5. Shipping Performance

```python
#Calculate the nummber of Shipping Days

df['Shipping Performance'] = pd.cut(
    df['Shipping Days'],
    bins=[-1, 2, 5, float('inf')],
    labels=['Fast', 'Standard', 'Slow'])
```

6. Returned Flag

```python
#Check whether the order returned or not

df['Returned Flag'] = (
    df['Returned']
    .map({'Yes': 1, 'No': 0}))
```

# **Exploratory Data Analysis(EDA)**

Exploratory Data Analysis (EDA) is the process of examining, summarizing, and visualizing a dataset to understand its main characteristics, identify patterns and trends, detect unusual values, and discover relationships between variables.

For this retail sales project, EDA was performed after data cleaning and feature engineering. The analysis helps understand sales performance, customer segments, product categories, payment methods, order trends, returns, and regional performance.

1. Descriptive Analysis

Calculated statistical summaries such as count, mean, median, minimum, maximum, and standard deviation for numerical variables such as Sales, Quantity, List Price, and Customer Feedback Score.

2. Univariate Analysis

Studied individual variables separately using histograms, bar charts, pie charts, and box plots. This helped understand the distribution of sales, quantities, categories, payment modes, and other variables.

3.Bivariate Analysis

Examined relationships between two variables, such as:

*   Segment vs Sales
*   Category vs Quantity
*   Payment Mode vs Sales
*   Order Year vs Sales
*   Returned vs Sales

4. Multivariate Analysis

Analyzed multiple variables together to identify more complex patterns, such as:
*   Category + Region + Sales
*  Order Year + Category + Sales

5. GroupBy Analysis

Used groupby() to calculate aggregated metrics such as total sales, average sales, quantity, and order counts across categories, regions, segments, and payment modes.

6. Pivot Table Analysis

+1 → Perfect positive relationship

0 → No linear relationship

-1 → Perfect negative relationship

The values range from -1 to +1, where positive values indicate positive linear association and negative values indicate negative linear association. # **7.Sales Share by Category**

Purpose

*The purpose of analyzing Sales Share by Category is to understand how much each product category contributes to the total sales of the retail business.*

*A pie chart is suitable here because it shows the proportion of each category as a part of the overall sales.*

```python
# Group the data by Category and calculate the total Sales for each category
category_sales = df.groupby('Category')['Sales'].sum()

# Generate a different color for each category using the Set2 color palette
colors = sns.color_palette('Set2', len(category_sales))

# Define a function to display percentage labels only when the share is 5% or more
def show_percentage(pct):

    # Display the percentage with one decimal place if it is >= 5%; otherwise display nothing
    return f'{pct:.1f}%' if pct >= 5 else ''

# Create the figure and pie chart area with a size of 9 × 7 inches
fig, ax = plt.subplots(figsize=(9, 7))

# Create the pie chart using category-wise sales
wedges, texts, autotexts = ax.pie(

    # Use total sales of each category as the pie chart values
    category_sales,

    # Apply the custom function to display percentage labels
    autopct=show_percentage,

    # Start the first slice from the 90-degree position
    startangle=90,

    # Apply the selected colors to the pie slices
    colors=colors,

    # Position the percentage labels closer to the center of the pie
    pctdistance=0.70,

    # Add white borders between slices for better visual separation
    wedgeprops={
        'edgecolor': 'white',
        'linewidth': 1.5
    })

# Loop through all percentage labels
for autotext in autotexts:

    # Set the percentage label font size to 11
    autotext.set_fontsize(11)

    # Make the percentage labels bold for better readability
    autotext.set_weight('bold')

# Add a legend to identify each category
ax.legend(

    # Use the pie chart slices as legend markers
    wedges,

    # Use category names as legend labels
    category_sales.index,

    # Add a title to the legend
    title='Category',

    # Position the legend on the left side of its anchor point
    loc='center left',

    # Place the legend outside the pie chart
    bbox_to_anchor=(1, 0.5))

# Add a title to the chart
ax.set_title(

    # Set the chart title
    'Sales Distribution by Category',

    # Set the title font size
    fontsize=16,

    # Make the title bold
    fontweight='bold',

    # Add spacing between the title and chart
    pad=20)

# Ensure the pie chart remains perfectly circular
ax.axis('equal')

# Automatically adjust the layout to prevent overlapping elements
plt.tight_layout()

# Display the final chart
plt.show()
```

# **13.Correlation**

Correlation is used to measure the strength and direction of the relationship between two numerical variables.

correlation can helps to understand relationships between variables such as:

2.Quantity

5.Shipping Days


```python
# Select the numerical columns that will be used for correlation analysis
correlation_columns = [
    'List Price',
    'Quantity',
    'Sales',
    'Customer Feedback Score',
    'Shipping Days']

# Calculate the correlation between all selected numerical variables
# .corr() calculates Pearson correlation coefficients
correlation_matrix = df[
    correlation_columns
].corr()

# Create the figure and set its size
plt.figure(figsize=(10, 7))

# Create a heatmap to visualize the correlation matrix
sns.heatmap(
     # Use the calculated correlation matrix as the data
    correlation_matrix,
    # Display the correlation values inside each cell
     annot=True,
    # Display correlation values with 2 decimal places
     fmt='.2f',
     # Apply the coolwarm color scheme
    # Different colors help identify positive and negative relationships
     cmap='coolwarm',
     # Add space between the cells for better readability
     linewidths=0.5,
      # Display the color bar showing the correlation scale
     cbar=True)

# Add a title to the heatmap
plt.title('Correlation Heatmap of Numerical Variables')
plt.xlabel('Variables')
plt.ylabel('Variables')

# Automatically adjust the layout to prevent overlapping elements
plt.tight_layout()
plt.show()
```
